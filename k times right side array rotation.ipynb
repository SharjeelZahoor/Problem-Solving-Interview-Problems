Rotate the list to the left by k steps.

give the list members according to the index given in queries list

```python
def circularArrayRotation(list1, k):
  n = len(list1)
  lst2=[]
  k = k%n
  for i in range(n-k, n):
    lst2.append(list1[i])
  for i in range(n-k):
    lst2.append(list1[i])

  return lst2



list1 = [2, 4, 6, 8, 10, 12, 14]
k = 10
circularArrayRotation(list1, k)
```
