check is there any palindrome in the list?

```python
def is_palindrome(lst):
    count = 0
    for word in lst:
      start = 0
      end = len(word)-1
      while start != end:
        if word[start] != word[end]:
          count+=1
          break
        start+=1
        end-=1
    if count == len(lst):
      return "no, there is not any plaindrome in the list"
    return "yes there is a plaindrome in the list"


lst = ["sir", "madam", "ali", 'zahra']
result = is_palindrome(lst)
print(result)



```
