Move all zeros to end of the list



```python
def findMaxConsecutiveOnes(nums):
  ans = 0
  count = 0
  for i in range(len(nums)):
    if nums[i] == 1 :
      count+=1
    if ans < count:
      ans = count
    if nums[i] != 1:
      count=0
  return f'Maximum number of consecutive ones are : {ans}'


nums = [1,1,2,3,1,1,1,1,2,3,1]
findMaxConsecutiveOnes(nums)
```
