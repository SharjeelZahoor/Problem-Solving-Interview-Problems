Rotate the list to the right by k steps.

give the list members according to the index given in queries list

```python
def circularArrayRotation(list1, k):
    n = len(list1)
    k = k % n
    rotated_array = list1[-k:] + list1[:-k]
    print(rotated_array)



list1 = [1,2,3,4,5,6,7]
k = 9
circularArrayRotation(list1, k)
```
