Move all zeros to end of the list



```python
def zerosAtEnd(lst):
  lst2=[]
  count=0
  for i in range(len(lst)):
    if lst[i] == 0:
      count+=1
    else:
      lst2.append(lst[i])
  for j in range(count):
    lst2.append(0)
  return lst2

lst=[1,3,2,0,5,0,4,0]
zerosAtEnd(lst)
```
