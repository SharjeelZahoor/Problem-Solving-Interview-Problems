Move all zeros to end of the list



```python
def zerosAtEnd(lst):
  i = 0
  j = len(lst)-1
  while i <= j:
    if lst[j] == 0:
      j-=1
    if lst[i] == 0:
      lst[i], lst[j] = lst[j], lst[i]
    else:
      i+=1
  return lst


lst=[1,3,2,0,5,0,4,0]
zerosAtEnd(lst)
```
